```python
import pandas
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

path = pandas.read_csv("emails.csv")
print("Path to dataset files:" , path)

path.head()
```

```python
# Check all values
print(path.isna())
```

```python
print(path.isna().sum())
```

```python
# checking duplicate rows
path.duplicated()[path.duplicated()==True]
```

```python
# finding number of rows and columns

print("Shape (rows, columns): ", path.shape,"\n")

print("number of rows: ", path.shape[0])
print("number of columns: ", path.shape[1])
```

```python
# viewing the data types of columns
path.dtypes
```

```python
# Statistical summary
path.describe(include='all')
```

```python
#Spam vs Not Spam Count (Bar Chart)
plt.figure(figsize=(6,4))
sns.countplot(x="spam", data=path)

plt.title("Spam vs Non-Spam Emails")
plt.xlabel("Spam (0 = No, 1 = Yes)")
plt.ylabel("Count")
plt.show()
```

```python
#Spam Distribution (Pie Chart)
path["spam"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    labels=["Not Spam", "Spam"],
    figsize=(6,6)
)

plt.title("Spam Email Distribution")
plt.ylabel("")
plt.show()
```

```python
Correlation Matrix
```

```python
plt.figure(figsize=(6,4))
sns.heatmap(path.corr(numeric_only=True), annot=True)

plt.title("Feature Correlation")
plt.show()
```
